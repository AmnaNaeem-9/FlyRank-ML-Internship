## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
# Task type: Classification

# I want to identify content items that are likely to be declining so that content editors can decide which items need attention first. This is a classification problem because the target is a yes/no outcome: whether a content item is labeled as declining. The model's output would support a content editor in deciding which content should be reviewed or improved.
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
# Target: trend_direction
# The target is trend_direction, which records the observed performance trend of each content item. It contains five categories: down, stable, up, new, and flat. The model would predict this category from other available content and performance signals. trend_pct will not be used as an input feature because trend_direction is derived from it, which would cause target leakage
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# Primary metric: Recall

# I will use recall as the primary metric because the goal is to identify as many genuinely declining content items as possible. A false negative means that a declining content item is not surfaced for editorial review. I will also examine precision because too many false positives could cause editors to spend time reviewing content that does not actually need attention.
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# A fixed rule such as "flag content when CTR is below a chosen threshold" would not capture the different signals associated with content performance trends. Content performance can depend on multiple factors such as engagement, search position, content type, traffic, content age, and other available signals. These signals can interact in ways that are difficult to represent with a small set of manually written rules. ML can learn patterns from historical content outcomes and combine multiple signals to predict the likely trend direction. The goal is to provide decision support for content editors rather than replace their judgment.
```
